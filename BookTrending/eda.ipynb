```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("atlantic_books_public.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
df["actual_price"].hist(bins=30)
plt.xlabel("Actual Price")
plt.ylabel("Frequency")
plt.title("Distribution of Actual Prices")
plt.show()
```

```python
plt.boxplot(df["actual_price"])
```

```python
df[df["actual_price"]>3000]
```

```python
plt.boxplot(df[df["actual_price"]<2000]["actual_price"])
plt.xlabel("Actual Price")
plt.ylabel("Count")
plt.title("Distribution of Book Prices")
```

```python
df["format"].value_counts().plot(kind="bar")
plt.xlabel("Format")
plt.ylabel("Count")
plt.title("Distribution of Book Formats")
```

```python
df["writer"].value_counts().sort_values(ascending=False).head(10)
```

```python
plt.pie(df["subject"].value_counts(), labels=df["subject"].value_counts().index, autopct='%1.1f%%')
```

```python
df.groupby("publisher", as_index=False)["actual_price"].mean()
```

```python
median_price = df["actual_price"].median()

df["price_category"] = df["actual_price"].apply(
    lambda x: "Budget" if x <= median_price else "Premium"
)
```

```python
df["price_category"].value_counts().plot(kind="bar", color=["green", "orange"])
plt.xlabel("Price Category")
plt.ylabel("Count")
plt.title("Distribution of Book Price Categories")
```
